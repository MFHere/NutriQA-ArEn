#### last all files in new folder:

```python
import os
import pandas as pd

def read_and_concat_files(folder_path):
    """
    Reads all .xlsx and .csv files in the specified folder and 
    concatenates them row-wise into a single DataFrame.
    
    Parameters:
        folder_path (str): The path to the folder containing the files.
        
    Returns:
        pd.DataFrame: A DataFrame containing all concatenated data.
    """
    all_dataframes = []

    for filename in os.listdir(folder_path):
        file_path = os.path.join(folder_path, filename)

        if filename.endswith('.csv'):
            try:
                df = pd.read_csv(file_path)
                all_dataframes.append(df)
            except Exception as e:
                print(f"Error reading CSV file {filename}: {e}")
                
        elif filename.endswith('.xlsx') or filename.endswith('.xls'):
            try:
                df = pd.read_excel(file_path)
                all_dataframes.append(df)
            except Exception as e:
                print(f"Error reading Excel file {filename}: {e}")
    
    if all_dataframes:
        return all_dataframes,pd.concat(all_dataframes, ignore_index=True)
    else:
        print("No valid CSV or Excel files found.")
        return pd.DataFrame()
```
